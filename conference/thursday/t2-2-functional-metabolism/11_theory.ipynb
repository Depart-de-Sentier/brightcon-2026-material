# Theory outline

Research question: What (novel) practitioner choices are required to deal with product lifetimes in time-explicit LCA, how should we grapple with these choices, and to what extent can we rely on existing tools and concepts?

Co-occurring discussions:
- The development of tools for time-explicit LCA (notably: `bw_timex` and `Trails`) and how the parallel conceptual development we need to apply them sensibly.
- Accounting of product systems in a Digital Product Passport (DPP): what activities to attribute to a particular product (and how).
- Analysis to support environmental governance (e.g. for a circular economy or Safe-and-Sustainable by Design (SSbD) innovations): the timing of material flows and technology adoption affects efficacy.

Objective:
- Develop an attributional approach to (deep) temporal differentiation (\& temporal evolution) of product lifetimes with a time-specific physical-flow logic. 

## Concepts that already have a broadly\*-agreed-upon meaning
<sup>*Relatively speaking. Your mileage may vary.</sup>

### Attributional perspective: a functional network

Here, we'll consider LCA in the attribution sense. This makes LCA an exercise in understanding society's metabolism as a functional network, where the provision of one function often depends on the provision of many other functions (e.g., obtaining a bicycle requires that the aluminium contained therein was somehow collected and shaped). This perspective traces goods 'upstream' to a process which provides this good and a waste is traced 'downstream' to a process which treats this waste.

When a process provides a good or treats a waste, this is a functional flow of the process. Logically, any process must have at least one functional flow, or it would have no reason to exist. When a process provides multiple functions, we speak of multifunctionality. This is a topic we'll not be getting into here.

Because the scale at which an activity occurs in reality rarely relates to the scale at which its function is consumed by a product system, we devise ways of attributing some of this activity to some of its function. To this end, we typically assume a linear relationship between the activity and its function.

<img src="figures_external/alca_in_brief.png" width="700"/>

### Temporalization

Conventionally, an LCA system represents a steady state, reported as being representative of some timespan. To the degree that any process of such a system can be generalised (i.e., assuming that it hasn't been constructed with the aim to break from its conceptual foundation), any time-dependent characteristic it may have has been flattened away, ironed into a steady state. We can argue that this approach is what has allowed us to conduct LCA in the first place, but since the early beginnings of this method, there have been those arguing for approaches which can be broadly labelled as dynamic LCA. 

The term 'dynamic LCA' has seen many uses. To more precisely describe what we're talking about here, we use the term 'temporal differentiation'. Considering the conventional matrix structure of LCA ([Heijungs & Suh, 2002](https://link.springer.com/book/10.1007/978-94-015-9900-9)), temporal differentiation refers to a technology matrix which has been expanded so that a particular process can be specified as occurring at one time or another. The elements of this matrix can then connect a product which one process creates at some time to the consumption of this product at another point in time. [Müller et al. (2025)](https://link.springer.com/article/10.1007/s11367-025-02539-3) and [Sporchia et al. (2026)](https://link.springer.com/article/10.1007/s11367-026-02721-1) each describe such an approach. Of course, temporal differentiation also has implications for the intervention matrix and the subsequent LCIA phase. These are outside of our present scope.

<img src="figures_external/temporalization_general.png" width="700"/>

At present, adding temporal differentiation to inventories rarely happens by explicitly defining the timesteps which each flow connects. Instead, we define a relative timing between the process that consumed the flow and the process that provides it (see [Temporalis](https://temporalis.readthedocs.io/en/latest/) by [Cardellini et al., 2018](https://www.sciencedirect.com/science/article/abs/pii/S0048969718325257)).

<img src="figures_external/temporalization_relative.png" width="700"/>

By defining temporal differentiation in terms of the relative timing between processes, the resulting datasets become applicable to a variety of use cases. Having defined the function which maps t<sub>a</sub> to t<sub>b</sub>, the same dataset can be used whether t<sub>a</sub> is 2000 or 2020 (limited by the temporal relevance of the inventory). In short, what would be many time-specific processes in the conventional (pre-determined) matrix structure can be encapsulated in a single 'process' and only expanded to time-specific processes as needed.

The benefits of this approach are best understood in relation to the concept of temporal evolution. Possibly, a process *occurring at* t<sub>x</sub> also has some implications for its *quantification*. If the process were to occur instead at t<sub>y</sub>, we could quantify it differently. The realisation of this distinction is the reflection of temporal evolution. Both of these aspects of temporalization are present in time-explicit LCA (see [Müller et al., 2025](https://link.springer.com/article/10.1007/s11367-025-02539-3)).

There are various ways of implementing temporal evolution: being able to map various processes as representing the same activity, but occurring at different times, or expanding a dataset with variables to directly adjust how the inventory is quantified based on what time (or vintage) it is called to represent. We'll get further into this later.

At the risk of getting ahead of ourselves, temporalization is most interesting when it allows us to easily represent diverse the diversity of objects which enable a particular use function. For example, global shipping is enabled by a heterogeneous collection of vessels (varying in size, speed, engine, etc.), which can be described over time based on the vintage composition of the in-use fleet.

<img src="figures_external/temporalization_relative_fleet.png" width="700"/>

#### Deep temporalization

When we speak of 'deep' temporalization, this reflects a systematic attempt to temporalize every (relevant) part of the product system, including what typically constitutes the background. Temporal differentiation at this level has been demonstrated by [DyPLCA](http://dyplca.univ-lehavre.fr/) (see [Pigné et al., 2020](https://link.springer.com/article/10.1007/s11367-019-01696-6)) and is also the aim of [Trails](https://github.com/Laboratory-for-Energy-Systems-Analysis/trails), which notably enables temporal evolution too.

What we are most interested in here is the development of such capacities, considering both temporal differentiation and evolution in a way that is conceptually consistent.

#### Shallow temporalization

Also known as foreground-only temporalization. A precise definition of 'shallow' temporalization is not necessary here.

What *is* important is to point out is that shallow temporalization may avoid challenges of deep temporalization all together. Shallow temporalization may permit itself to only consider products the lifetime of which is fully encapsulated by the reference flow, thereby neglecting the need to implement a physical-flow logic that is explicitly time-specific. Or, a dynamic stock model (DSM) may be used to depart from a lifecycle physical-flow logic all together. Both of these concepts are explained below.

In short, while shallow temporalization may align (to a limit extent) with a time-specific physical-flow logic, this alignment can avoid the challenges of deep temporalization, which is a shortcut that we're not interested in indulging here.

### Lifetime terminology

Some of the basic techniques involved in creating a DSM are outlined in the `01` notebook. Here, a brief overview of how we use some key terms and their meaning within the case at hand.

- **Age-cohort**: We can group objects together based on the time of their production and associate various traits to it (e.g., at comparable storage volumes, a fridge produced in 1980 uses more material and consumes more electricity than a fridge produced in 2010). In our usage, all members of an age-cohort share a unique vintage.
- **Inflow**: This is the addition of items to a stock across a particular interval. Here, we consider any inflow to result from a production activity, assuming that any object enters the stock of in-use products as soon as it is produced.
- **Outflow**: This is the removal of items from a stock across a particular interval. The age-cohort approach used here means that the total outflow can be disaggregated according to its age-cohort composition. Here, we consider outflow to result in a disposal activity, assuming that any object is disposed of as soon as it leaves the stock of in-use products.
- **Survival**: The data we use here (see `01` notebook) assumes that a given technology (ICEV or BEV) has a set survival function, regardless of its vintage. This means that for each cohort, the relationship between its retired and remaining fractions is entirely determined by its age (cf. [Krych et al., 2024](https://onlinelibrary.wiley.com/doi/10.1111/jiec.13586)). Broadly, there are many characteristics of product lifetime which are here simplified or out of scope (cf. [Jerome & Ljunggren, 2025](https://link.springer.com/article/10.1007/s11367-025-02470-7)).
- **Stock**, **use function**, and **in-use consumption**: From their inflow up to their outflow, objects are counted as part of the stock. For each interval, the change in stock is equal to the difference between its inflow and outflow. Here, we consider the stock of objects to represent the in-use fleet. In our LCA approach, there is no economic flow which represents the stock itself. However, the use of the stock and its related in-use consumption are represented as economic flows, which could not be determined without knowledge of the stock. In this case (see `01` notebook), the use of the stock is entirely a function of time (specific use is expressed in km per vehicle per year), while the associated energy consumption is a function of vintage as well as technology (specific energy consumption is expressed in MJ per km per year). Note that these assumptions do not allow for dormant or redundant stocks (cf. [Van Engelenburg et al., 2026](https://iopscience.iop.org/article/10.1088/2634-4505/ae9642)). 
- **Throughput**: It is possible for objects to leave the stock within the same interval that they entered the stock. This leads to a quantity of inflow and outflow that never registers as part of the stock. Such a throughput is not adequately attributed following our present approach, as we attribute the production and disposal of objects based on the use they provide.

## Concepts which are novel or have a poorly-defined meaning

### Part 1: Logics to defining a time-explicit process

#### Physical-flow logic

When engaging with the concept of temporal differentiation, this is commonly thought of as representing some 'time dependency' (see, e.g., [Tiruta-Barna et al., 2016](https://www.sciencedirect.com/science/article/pii/S0959652615018739)). For attributional LCA, this means that to realise a particular function at a particular time, we must investigate the timeline for activities which (in some part) function to (directly) enable it. For example, the enjoyment of a cup of take-away tea becomes responsible for some harvest of tea leaves, but also for the disposal of a tea cup. Following a time-dependency logic, the *what and when* of this harvest and this disposal is determined by connecting these functions to resemble how their physical flows can be traced. We'll refer to this here as a 'physical-flow logic'.

An exhaustive implementation of such a functional web would necessarily stretch back to the first peoples to use technology, as well as stretching forward to the last peoples who will do so. That this isn't practically feasible should not deter us, although it should lead us to reevaluate certain concepts. We'll mostly leave this as a matter for another time. Notice, however, the distinction with consequential LCA: as a consequence cannot precede its cause, a CLCA system only stretch forwards from the time of the evaluated decision (see, e.g., [Schaubroeck et al., 2021](https://www.mdpi.com/2071-1050/13/13/7386)).

**Time-generic variant:** Understanding the interconnection of activities as stemming from physical flows, the question remains how to attribute the lifetime function provided by a product to the demand for this function at any time and quantity for which it is provided. One approach is to not consider any time-dependency between the activities enabling this lifetime and the moments at which its function is provided. This way, the demands associated across the entire lifetime are linearly scaled, regardless of the time at which its function is requested.

<img src="figures/time_generic_physical_flow.png" width="600"/>

**Time-specific variant:** An alternative to this time-generic approach is a deeper consideration for time-dependency, which takes the position that the consumption enabling functional provision at some time should also be attributed to the function provided at that time in particular. In our present example, the use of the vehicle in 2030 is associated with a corresponding energy demand in 2030 alone. Its energy demand in other years is thereby attributed to its use in those years.

<img src="figures/time_specific_physical_flow.png" width="600"/>

**Further sub-variants:** To some, this time-specific approach is the more intuitive way of attributing these demands to the function provided at a particular time. However, it also further complicates how these demands should be attributed. Both the time-generic and time-specific cases demonstrated above scale the flows of the object (production and disposal) based on an equal weighting of the function provided across the object's full lifetime. Because, additionally, this case doesn't consider any change in the object's in-use efficiency across its lifetime, summing the first-tier demands over time yields the same quantities for each case.

However, different assumptions could have been made, pointing to the existence of additional sub-variants. One alternative is to not weight every instance of functional provision equally, but, for example, to give the first 10\% of the lifetime function a higher weight than the next 10\%. Another alternative is to perform the initial attribution of the object according to its lifespan, so that each timestep in which some of its function is provided has a particular (equal or otherwise defined) weight, regardless of the scale of the function provided in that timestep. Regardless of what particular motives may exist to opt for such alternatives, we do not consider them any further here.

#### Alternatives to physical-flow logic

A brief detour before we move on. The functions enabling the lifetime of a product don't have to be defined (conceptually or practically) by the physical flows associated with *this particular* product. Here, we discuss a few alternatives to the physical-flow logic which define what flows to connect based on their placement in time. We therefore provisionally label it here as a temporal-flow logic.

**Unit-based co-occurring temporal-flow logic:** The most frequently occurring alternative is to associate the use of a product with the production and disposal of this product (or similar products) at the time when this use occurs. For example, the use of a bike in 2026 may be modelled as requiring the production of a bike in 2026 and the disposal of a bike in 2026 (using data specific to how these activities occur in that year). This is effectively what is typically assumed in static LCA.

<img src="figures/unit_based_co_occurring_temporal_flow.png" width="600"/>

**Activity-based co-occurring temporal-flow logic:** An alternative approach to the above example is not to scale these activities to a function expressed in terms of a set number of products (i.e., where the full use of 1 product requires that exactly 1 product enters the use phase), but a function expressed in terms of an elapsed time. This way, the use of bikes in 2026 would be quantified with the respective production and disposal of bikes in 2026. There is no intrinsic need for the quantity of bikes embodied in these functions to align with each other (e.g., if the stock of bikes increased in 2026, more bikes were produced than were discarded). This approach is often implied when a DSM is used to report impacts occurring in a particular year.

<img src="figures/activity_based_co_occurring_temporal_flow.png" width="600"/>

**Activity-based weighted-timespan temporal-flow logic:** For both of the above alternative logics (broadly, anything that can be described as a 'co-occurring temporal-flow logic'), applying them consistently for a functional unit which is confined to a single timestep would mean that only this time step occurs within the LCA product system -- and, therefore, feature no temporal differentiation. However, temporal differentiation would occur for a functional unit which spans multiple timesteps or an assessment which considers multiple product systems over time.

At the same time, the conception of attribution underpinning these logics (that tracing physical flows is not required for attribution) may be used to develop innumerable approaches which do feature internal temporal differentiation. For example, rather than attributing the production of bicycles in 2026 to 2026 alone, this year's production activities may be attributed for 40\% to 2026, 30\% to 2027, 20\% to 2028, and 10\% to 2029. The number of years selected and how their distribution is determined could follow any of various lines of reasoning. For example, following the logic with which the depreciation of assets is allocated in accountancy. This may be an attractive approach when no data can be feasibly made available to support a physical-flow logic (or, perhaps, when it is undesirable to match the functioning of a system to discrete physical flows).

<img src="figures/activity_based_weighted_timespan_temporal_flow.png" width="600"/>

### Part 2: Defining a time-explicit process beyond a single-object perspective

Having defined the time-specific physical-flow logic, but only demonstrated it for a very simple case: an object with a set production year and disposal year. Indeed any particular object typically does have a set lifespan. However, we rarely have both the capacity and the desire to model a product system as a network of particular objects, because the number of objects involved is too great. Instead, we make generalising assumptions in order to reflect a variety of objects with just a single process.

One such generalising assumption is the definition of an age-cohort (also called a vintage, defined above). Acknowledging that such a cohort can include objects with diverse lifespans, the associated disposal activities are spread over time.

<img src="figures/time_specific_physical_flow_cohort.png" width="600"/>

This is where we start to get into trouble. These disposal activities are not a characteristic of the age-cohort as a whole (considering every object that could at any time be counted as one of its members) but specifically those members which are active at the time of use. Furthermore, recall that the quantity of an object attributed to use is here proportional to the use delivered over its lifetime. This means that those members of the cohort which retire shortly after a particular use is delivered require a larger quantity of the object than those members of the same age-cohort which retire far beyond the moment of this particular use.

Extending the definition of a functional demand from the function of an object to that of an age-cohort, we can extend it further to the function of the entire active fleet.

<img src="figures/time_specific_physical_flow_fleet_single_tech.png" width="600"/>

This step does not (much) worsen the trouble we're in -- the relationship between any individual cohort called above to its cohort-specific disposal functions the same way. To this end, be aware that the depiction of the demand for 'disposal' above does not exhibit this distinction between cohorts. This distinction is represented below.

### Part 3: Logics to defining a time-explicit functional unit

Although somewhat removed from our current focus, let's take a moment to consider the above concepts affect the definition of an assessment's functional unit.

The functional unit is the assessment's comparative basis. Conventionally, the reference flow which fulfils a function unit is expressed as a final demand vector. Often (but by no means necessarily), this demand vector has just one non-zero coefficient, representing a single economic flow. In cradle-to-grave studies, this flow typically encapsulates a use function. 

Following a physical-flow logic, such a reference flow would encapsulate the cradle-to-grave function provided by an object over its full lifetime.

<img src="figures/lifecycle_individual.png" width="600"/>

However, it may be just as valid to define a functional unit that requires the full use functions provided across an entire age-cohort. This could be defined as the full function, or scaled to some particular parameter. Such a function unit answers a question relating to this age-cohort in particular. How this cohort's existence relates to past and future cohorts is not captured.

<img src="figures/lifecycle_cohort.png" width="600"/>

An extension of this idea is to define the functional unit as relating to some set of age-cohorts. Although the precise set considered is likely to be arbitrary, this approach allows the practitioner to consider the object of analysis in a broader context.

Note that for each of these cases, the choice between a time-specific or time-generic approach to the process(es) providing the use flow(s) is of no consequence. Assuming that at most cohort-specific processes are defined, a time-generic approach could allow the first two cases to define a demand vector with a single non-zero coefficient. The third case, however, would require as many non-zero coefficients as the age-cohorts considered. On the other hand, a time-specific approach cannot define the full lifetime use function in a single coefficient, but so must include non-zero coefficients not just for the cohorts considered, but for each cohort at each timestep. 

<img src="figures/lifecycle_cohorts.png" width="600"/>

Considering how time-explicit LCA can be used to inform eco-design, we have introduced the labels 'metabolic time-explicit LCA' and 'mosaic time-explicit LCA' in a manuscript under review, [available as preprint](https://doi.org/10.21203/rs.3.rs-9554797/v1). These terms reflect how the assessment's functional unit is defined. The first two examples given here classify as 'mosaic' and the third as 'metabolic'. In this manuscript, we argue that the context which eco-design should consider is rarely limited to a single object or age-cohort and that, when this is the case, time-explicit LCA is wasted on a mosaic functional unit.

Note that these terms are defined in the context of a particular way of considering cradle-to-grave lifecycles and time-explicit LCA. Even while maintaining an understanding of a lifecycle as representing lifetime use function, these three cases are a limited selection of all possible approaches. Instead of defining a single cohort as having a shared production year, one could consider the lifetime of all objects disposed of in a certain year, or all objects active in a certain year. Instead of considering a use function scaled to the involved cohorts, it could be scaled to a particular total function (e.g., driving a total of 100 km) or to a particular number of objects (e.g., distance driven by 10 cars, each with a particular lifetime). These different ways of defining the form and scale of the functional unit lend themselves to a variety of different research questions.

## Realising a time-specific physical-flow logic

Now that we understand what we mean by this logic, let's consider some requirements for its implementation.

1. To be able to link production and disposal processes so that these reflect temporal evolution (e.g., vintage-specific attributes).
2. To be able to assign temporal distributions to production and disposal processes so that these occur at the appropriate time with respect to the time at which use occurs.
3. To be able to represent stocks with diverse vintage compositions (e.g., skewing younger, skewing older, excluding vintages before or beyond a certain year, ...).

Next, let's have a look at emerging tools to see what techniques we have at our disposal.

### Current tools

To consider the capabilities at our disposal, we'll consider two tools which are under active development: [`bw_timex`](https://docs.brightway.dev/projects/bw-timex/en/latest/) (see [Diepers et al., 2026](https://joss.theoj.org/papers/10.21105/joss.09621)) and [`Trails`](https://github.com/Laboratory-for-Energy-Systems-Analysis/trails).

#### Outline of `bw_timex`

Using `bw_timex`, there are three main ways of representing temporal information.

**1. Defining a `TemporalDistribution` and using `add_temporal_distribution_to_exchange` (or `exchange_object['temporal_distribution']`)**

By using a `TemporalDistribution`, the user defines a *relative* timing: either between the 'consumer' and 'producer' of an economic flow or the 'producer' of an environmental flow and when this flow occurs in the biosphere.

As a name suggests, this relative timing does not have to be a set duration (e.g., 2 years ago), but can describe some distribution over time (e.g., 20\% today and 80\% tomorrow).

**2. Defining `temporal_evolution_factors` or `temporal_evolution_amounts` and using `add_temporal_evolution_to_exchange`**

This allows you to define how the flows of a process differ based on whether it occurs at one time or another. By defining these factors/amounts in calendar time, the process becomes a function of the time at which it is created to occur.

**3. Defining `representative_time` and using `set_database_metadata`**

It is possible to assign a (background) database to a particular `representative_time`. This way, when a product from such a database is consumed at a particular time, `bw_timex` locates its producing process in the database that matches the time at which this product aught to be consumed. If there are multiple matches, these may be combined in a temporal market (e.g., if a product is required in 2024, this may be quantified as 60\% of the 2020 process with 40\% of the 2030 process).

#### What `Trails` adds

The `Trails` library has some features in common with `bw_timex`, but there some big differences.

`Trails` considers some set of time-specific technology matrices and interpolates these to an annual resolution. This is conceptually similar to the `representative_time` approach of `bw_timex` with its temporal markets. However, its approach to temporal differentiation is strictly annual. It also performs its calculations differently: instead of creating a single time-expanded technology matrix, it performs `temporal_routing`. This creates a time-indexed graph that, once certain criteria are reached, quantifies its remaining demands with the (static) year-specific matrices. This means that multiple calculations of the same system don't necessarily provide the same results, but that deep temporalization becomes possible at a much lower computational burden.

`Trails` relies on a particular process-product combination being present in all relevant databases. It has no analogue to the `temporal_evolution_factors` of `bw_timex`. However, its implementation of temporal evolution does consider something else, called `temporal_amount_source`. Usually, assigning a temporal distribution to an exchange simply describes how a set quantity should be distributed over time. However, it may be more appropriate if this value was instead recalculated based on the equivalent exchange quantity in the target years, which they refer to as matrix-stored values. For example, a process (called to occur in year t<sub>x</sub>) may consume 2 kWh electricity, but assign this a temporal distribution which covers t<sub>y</sub>, where the equivalent process consumes 1 kWh. Say that the temporal distribution from t<sub>x</sub> assigns a weight of 20\% to t<sub>y</sub>, then the usual (referred to as 'port amount') approach would quantify this as $2 \cdot 0.2 = 0.4$, while the matrix-stored approach would quantify this as $1 \cdot 0.2 = 0.2$. The utility of this feature is highly dependent on how unit processes have been configured and what they aim to represent.

### How unit processes can be constructed and configured

Here, we discuss the challenges these tools face in representing a time-specific physical-flow logic. We do this by considering the timing and temporal evolution of production and disposal processes with respect to a time-specific use demand. Note that extending this discussion to in-use consumption flows would lead to further complications, which we only briefly touch on here.

#### 1. Use to time-specific production and disposal

A 'use' process which calls production and disposal with temporal distributions based on a dynamic model.

This requires the disposal processes to have been pre-defined as reflecting a time-specific mix of vintages being disposed.

<img src="figures_external/configuration1.png" width="800"/>

Advantages:

- Process links are straightforward to interpret.
- Easy to verify relative timing and the mass balance of processes.

Trade-offs:

- Actually an activity-based weighted timespan temporal-flow logic in disguise: year-specific disposal processes reflect all disposal in that year, so also the disposal of objects which may not yet exist in the calling 'use' year.
- Definition of time-specific disposal processes occurs through an additional processes based on stock calculations. This forms an additional barrier to widespread usability.
- Lacking vintage-specific disposal processes, the effect of vintages on these flows is not readily interpretable.

#### 2. Use to time-specific production which in turn links to disposal

The 'use' process demands include flows which represent both production and disposal, effectively leading to a 'lifecycle excluding use' process. There are several ways of implementing this. Below, we illustrate disposal as a pulse demand at a set timestep following production. This could also be modelled as a distribution over time.

<img src="figures_external/configuration2.png" width="500"/>

Advantages:

- Vintage-specific disposal processes become possible, enabling richer interpretation (cf. first solution).
- The correct physical-flow-based vintage-specific quantity of production and disposal flows can be reflected (cf. first solution).
- The respective temporal distributions linking to production and disposal are intuitive to interpret and convenient to play around with and adapt to diverse applications. 
- This process configuration already frequently occurs in ecoinvent and premise-generated databases, making this a convenient solution to deep temporalization. 

Trade-off:

- As long as the aim is to create generalisable datasets (e.g., so that the 'lifecycle excluding use' process of a given vintage can be called from a variety of possible years), the temporal distribution of disposal cannot be defined in a way that consistently reflects its actual timing, which is co-defined by the surviving age-cohort when use occurs. If the aim of (deep) temporalization is to achieve conceptually consistent temporal differentiation and to be able to gain some meaning from the resulting timings, the mismatch which this solution creates between the timings of use related disposal activities is profoundly problematic.

#### 3. Use to time-and-age-specific production which in turn links to disposal

Being aware of the trade-offs of the above two solutions, it seems that the requirements we set out cannot be achieved if we're limited to defining a single dataset that covers each instance of production or disposal. Recall that the advantage of such a single dataset is that it can fulfil a role equivalent to many columns in the technology matrix.

Departing from this ideal, all sorts of alternative solutions become possible. Here we're considering just one of these, where collections of datasets for production (or rather, the lifecycle excluding use) are defined based on how much time has passed between the creation of a cohort and the time at which its use function is requested, with datasets for disposal being defined on a purely vintage basis, while considering the diverse temporalizations associated with the disposal of a single age-cohort.

<img src="figures_external/configuration3.png" width="500"/>

Advantages:

- Allows us to meet almost all of the requirements.
- In theory, this allows for robust, detail-oriented interpretation.

Trade-offs:

- This is difficult to set up and a pain to work with. The advantage of temporalization tools such as those of `bw_timex` is that many processes can be created from a single dataset. By requiring the creation of many such datasets, any use of the database becomes more unwieldy.
- To fully meet the requirements, this approach must allow a time-specific specification of temporal distributions, both between use and production as well as between production and disposal. This is because these distributions are entangled with how the stock has evolved up to the point of use. Ideally, we should be able to define a stock as having a larger share of younger cohorts at one time than another time. As far as we're aware, no tools for time-explicit LCA allow defining the temporal distributions of a dataset in this way.

#### Some notes on in-use consumption

Assuming that in-use consumption (and related economic and environmental flows) are also subject to temporal evolution, the related factors must either be embedded in the related 'use' process (as is done with 'disposal' in solution 1) or the 'production' process (as is done with 'disposal' in solution 2). These approaches lead to similar trade-offs regarding interpretation and timing respectively.

An additional challenge arises in correctly quantifying these consumption flows with respect to the object flows. Following our conceptualisation of time-specific physical-flow logic, providing a given function from two objects of the same vintage but with different lifetimes may lead to the same demands associated with in-use consumption, but to different demands for production and use: attributing these flows equally across the lifetime function provided means that the object with the shorter lifetime will require a larger quantity of these flows for the same use function.

Considering that age-cohorts feature diverse lifetimes, the production and disposal demands attributed to a given use function will differ for a cohort in its first year compared to the same cohort in a later year: the cohort operating in a later year only includes objects which have already passed a certain lifespan.

In short, this complicates the relationship between the demand quantity of the vintage-specific production and vintage-specific in-use consumption. Any approach to time-specific physical-flow logic must reckon with this.

### Constructing and configuring unit processes if we had additional capabilities

Recognising that each of the above approaches leads to less-than-ideal trade-offs, we are tempted to suggest that a preferred solution may entail expanding the capabilities with which time-explicit inventories can be defined. We provide one particular example, noting again that many alternative approaches could be devised.  

<img src="figures_external/suggestion.png" width="700"/>

Here, when use is described at occurring at a particular time, t<sub>b</sub>, relying on the use of any product active at t<sub>b</sub>, the temporal distribution of these production activities (the set of t<sub>a</sub>) is defined as a function of t<sub>b</sub>. Each of these resulting production processes (which describe a particular vintage created at a particular time) in turn calls the appropriate disposal activities. The time difference between use and production means that only the production of a share of the initial cohort is considered, namely, the share which is (still) active at t<sub>b</sub>. This affects the definition of the related disposal activities, as discussed above.

In this example, the dataset describing vintage-and-time-specific disposal does not need to know at what time use occurs. However, this information *is* required in order to accurately create and link processes from this dataset to the related production activities. Put differently, identifying the time and quantity of an object's disposal that follows from its particular use requires that this use is known, even though the object's use can otherwise be abstracted from its disposal.

This could be realised with the introduction of two new capabilities:

1. On-demand construction of a `TemporalDistribution` co-determined by when the connected process occurs as well as information passed along from an earlier computational step.
2. Exchange-specific factors akin to `temporal_evolution_factors` which are not determined by a specified time that is external to when the exchange is requested to occur. This time is thereby not defined in relative terms, but according to calendar time.
