# Presentation

Research question: What (novel) practitioner choices are required to deal with product lifetimes in time-explicit LCA, how should we grapple with these choices, and to what extent can we rely on existing tools and concepts?

Co-occurring discussions:
- The development of tools for time-explicit LCA and the parallel conceptual development we need to apply them sensibly.
- Accounting of product systems in a Digital Product Passport (DPP): what activities to attribute to a particular product (and how).
- Analysis to support environmental governance: the timing of material flows and technology adoption affects efficacy.

Objective:
- Develop an attributional approach to (deep) temporalization of product lifetimes with a time-specific physical-flow logic.
